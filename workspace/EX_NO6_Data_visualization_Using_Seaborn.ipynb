**Data visualization Using Seaborn**

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Datas
x = [1, 2, 3, 4, 5]
y = [3, 6, 2, 7, 1]
```

```python
plt.figure(figsize=(8, 5))
sns.lineplot(x=x, y=y)
plt.title('Simple Line Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y1 = [3, 5, 2, 6, 1]
y2 = [1, 6, 4, 3, 8]
y3 = [5, 2, 7, 1, 4]
```

```python
plt.figure(figsize=(10, 6))
sns.lineplot(x=range(1, 6), y=y1, label='y1', marker='o')
sns.lineplot(x=range(1, 6), y=y2, label='y2', marker='s')
sns.lineplot(x=range(1, 6), y=y3, label='y3', marker='^')
plt.title('Multiple Line Plot with Parameters')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid(True)
plt.show()
```

**Bar Plot - Seaborn**

```python
tips = sns.load_dataset('tips')
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='day', y='total_bill', data=tips, hue='sex', palette='Set2')

plt.xlabel('Day of Week')
plt.ylabel('Total Bill')
plt.title('Total Bill by Day with Gender Comparison')
```

```python
tit = pd.read_csv("datasets/titanic_dataset.csv")
tit
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Embarked', y='Fare', data=tit, palette='rainbow')
plt.title("Fare of Passenger by Embarked Town")
plt.xlabel('Embarked Town')
plt.ylabel('Fare')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Embarked', y='Fare', data=tit, palette='rainbow', hue='Pclass')
plt.title("Fare of Passenger by Embarked Town, Divided by Class")
plt.xlabel('Embarked Town')
plt.ylabel('Fare')
plt.show()
```

```python
import seaborn as sns
tips = sns.load_dataset('tips')
plt.figure(figsize=(8, 6))
sns.scatterplot(x='total_bill', y='tip', data=tips, hue='sex', s=100)

plt.xlabel('Total Bill')
plt.ylabel('Tip Amount')
plt.title('Scatter Plot of Total Bill vs. Tip Amount')
plt.show()
```

**VIOLIN PLOT**

```python
plt.figure(figsize=(10, 6))
sns.violinplot(x='Sex', y='Age', data=tit, palette='Set2')
plt.title('Distribution of Age by Sex (Violin Plot)')
plt.xlabel('Sex')
plt.ylabel('Age')
plt.show()
```

**HISTOGRAM**

```python
import seaborn as sns
import numpy as np
import pandas as pd
```

```python
np.random.seed(0)
marks = np.random.normal(loc=70, scale=10, size=100)
```

```python
# Generating dataset of random numbers
np.random.seed(1)
num_var = np.random.randn(1000)
num_var = pd.Series(num_var, name = "Numerical Variable")
num_var
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=marks, kde=True, bins=20, color='skyblue')
plt.title('Histogram of Marks')
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=tit, x='Pclass', hue='Survived', kde=True, stat='count', palette='husl')
plt.title('Passenger Class Distribution by Survival Status')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.show()
```
